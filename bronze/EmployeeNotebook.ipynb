```python
from pyspark.sql import SparkSession

# Start Spark session (already available in Databricks as 'spark')
# spark = SparkSession.builder.getOrCreate()

# Sample employee data
data = [
    (1, "Alice", "HR", 50000,"35","Hyderabad"),
    (2, "Bob", "IT", 60000,"45","Chennai"),
    (3, "Charlie", "Finance", 70000,"50","Hyderabad"),
    (4, "David", "IT", 65000,"40","Chennai"),
    (5, "Eva", "HR", 55000,"36","Delhi")
]

# Define schema (id, name, department, salary)
columns = ["emp_id", "name", "department", "salary","age","city"]

# Create DataFrame
df_employee = spark.createDataFrame(data, columns)

# Show DataFrame
df_employee.display()   # Databricks notebook
# or
df_employee.show()
# mycatalog.bronze
```

```python
df_employee.createOrReplaceTempView("vw_df_employee")
```

```python
df_final = spark.sql("""
          
          select emp_id, name, age,city, department, salary from vw_df_employee
          
          """)
```

```python
df_final.printSchema()
```

```python
df_final.write.mode("overwrite")\
        .option("overwriteSchema","false")\
        .option("mergeSchema","false")\
        .saveAsTable("mycatalog.bronze.tblEmployee_bronze")
```
