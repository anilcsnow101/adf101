```python
catalog = "mycatalog"
schema = "bronze"
table_name = "product_bronze"
```

```python
df_products = spark.read.format("csv") \
    .option("header", True) \
    .option("inferSchema", True) \
    .option("sep", ",") \
    .option("skipRows", 2) \
    .load("/Volumes/mycatalog/bronze/salesdata/productData.csv")
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
df_products_final = df_products.select(
    [col(c).alias(c.replace(" ", "_")) for c in df_products.columns]
)
df.printSchema()
if df_products_final.count() > 0:
   df_products_final.write \
      .format("delta") \
      .mode("overwrite") \
      .option("overwriteSchema", "true") \
      .saveAsTable(f"{catalog}.{schema}.{table_name}")
else:
    print(f"Product file is empty")   
  
```
